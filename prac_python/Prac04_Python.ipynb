Создайте класс прямоугольник — Rectangle. Метод __init__ принимает две точки — левый нижний и правый верхний угол. Каждая точка представлена экземпляром класса Point. Реализуйте методы вычисления площади и периметра прямоугольника.


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class Rectangle:
    def __init__(self, bottom_left, top_right):
        self.bottom_left = bottom_left
        self.top_right = top_right

    def width(self):
        return abs(self.top_right.x - self.bottom_left.x)

    def height(self):
        return abs(self.top_right.y - self.bottom_left.y)

    def area(self):
        return self.width() * self.height()

    def perimeter(self):
        return 2 * (self.width() + self.height())
```

Упражнение 3.
Добавьте в класс Rectangle метод contains. Метод принимает точку и возвращает True, если точка находится внутри прямоугольника и False в противном случае.


```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

class Rectangle:
    def __init__(self, bottom_left, top_right):
        self.bottom_left = bottom_left
        self.top_right = top_right

    def width(self):
        return abs(self.top_right.x - self.bottom_left.x)

    def height(self):
        return abs(self.top_right.y - self.bottom_left.y)

    def area(self):
        return self.width() * self.height()

    def perimeter(self):
        return 2 * (self.width() + self.height())

    def contains(self, point):
        return (self.bottom_left.x <= point.x <= self.top_right.x and
                self.bottom_left.y <= point.y <= self.top_right.y)
```

```python
p1 = Point(1, 1)
p2 = Point(4, 5)
rect = Rectangle(p1, p2)

point_inside = Point(2, 3)
point_outside = Point(0, 0)

print(rect.contains(point_inside))
print(rect.contains(point_outside))
```

Упражнение 4.
Описать класс десятичного счётчика. Он должен обладать внутренней переменной, хранящей текущее значение, методами повышения значения (increment) и понижения (decrement), получения текущего значения get_counter. Учесть, что счётчик не может опускаться ниже 0.


```python
class DecimalCounter:
    def __init__(self, initial_value=0):
        self._counter = max(0, initial_value)

    def increment(self):
        self._counter += 1

    def decrement(self):
        if self._counter > 0:
            self._counter -= 1

    def get_counter(self):
        return self._counter
```

```python
counter = DecimalCounter(5)
print(counter.get_counter())

counter.increment()
print(counter.get_counter())

counter.decrement()
print(counter.get_counter())

for _ in range(10):
    counter.decrement()
print(counter.get_counter())
```

Упражнение 6.
Доработать предыдущую задачу, чтобы можно было складывать двое часов друг с другом. Для перегрузки оператора + использовать метод __add__(self, other).


```python
class Clock:
    def __init__(self, hours=0, minutes=0, seconds=0):
        self.hours = hours % 24
        self.minutes = minutes % 60
        self.seconds = seconds % 60

    def add_hour(self):
        self.hours = (self.hours + 1) % 24

    def add_minute(self):
        self.minutes += 1
        if self.minutes >= 60:
            self.minutes = 0
            self.add_hour()

    def add_second(self):
        self.seconds += 1
        if self.seconds >= 60:
            self.seconds = 0
            self.add_minute()

    def __add__(self, other):
        total_seconds = (self.hours * 3600 + self.minutes * 60 + self.seconds +
                        other.hours * 3600 + other.minutes * 60 + other.seconds)

        hours = total_seconds // 3600 % 24
        minutes = (total_seconds % 3600) // 60
        seconds = total_seconds % 60

        return Clock(hours, minutes, seconds)

    def get_time(self):
        return f"{self.hours:02d}:{self.minutes:02d}:{self.seconds:02d}"
```

```python
clock1 = Clock(10, 30, 45)
clock2 = Clock(5, 45, 20)

clock3 = clock1 + clock2
```

Создать классы для травоядного животного и травы. Животное должно уметь поедать траву, если испытывает голод, в противном случае отказываться от лакомства. Трава должна обладать питательностью, в зависимости от которой животное будет насыщаться.


```python
class Grass:
    def __init__(self, nutrition):
        self.nutrition = nutrition

class Herbivore:
    def __init__(self, hunger_level=0):
        self.hunger_level = hunger_level

    def eat(self, grass):
        if self.hunger_level > 0:
            self.hunger_level = max(0, self.hunger_level - grass.nutrition)
            return True
        else:
            return False

    def get_hunger_level(self):
        return self.hunger_level
```

```python
animal = Herbivore(hunger_level=10)
grass1 = Grass(nutrition=3)
grass2 = Grass(nutrition=5)

print(animal.eat(grass1))
print(animal.get_hunger_level())

print(animal.eat(grass2))
print(animal.get_hunger_level())

grass3 = Grass(nutrition=10)
print(animal.eat(grass3))
print(animal.get_hunger_level())

print(animal.eat(grass1))
```

Упражнение 9.
Программа написана верно, однако содержит места потенциальных ошибок.
⦁	найдите потенциальные источники ошибок (укажите номера строк в строке документации);
⦁	используя конструкцию try добавьте в код обработку соответствующих исключений.

Потенциальные ошибки:

1.   48: accounts[account_from] - KeyError если аккаунт не существует
2.   49: value > balance_from - NoMoneyToWithdrawError если недостаточно
3.   50: accounts[account_to] - KeyError если аккаунт не существует
4.   51: accounts[account_from] = ... - общее исключение при операции
5.   52: accounts[account_to] = ... - общее исключение при операции
6.   53: value <= 0 - ValueError если сумма перевода не положительная



```python
class NoMoneyToWithdrawError(Exception):
    def __init__(self, message):
        super().__init__(message)


class PaymentError(Exception):
    def __init__(self, message):
        super().__init__(message)


def print_accounts(accounts):
    """Печать аккаунтов."""
    print("Список клиентов ({}): ".format(len(accounts)))
    for i, (name, value) in enumerate(accounts.items(), start=1):
        print("{}. {} {}".format(i, name, value))


def transfer_money(accounts, account_from, account_to, value):
    """Выполнить перевод 'value' денег со счета 'account_from' на 'account_to'
    """
    try:
        # Проверка на положительную сумму перевода
        if value <= 0:
            raise ValueError("Сумма перевода должна быть положительной")

        # Проверка существования счетов
        if account_from not in accounts:
            raise KeyError(f"Счет отправителя '{account_from}' не существует")
        if account_to not in accounts:
            raise KeyError(f"Счет получателя '{account_to}' не существует")

        balance_from = accounts[account_from]
        balance_to = accounts[account_to]

        # Проверка достаточности средств
        if value > balance_from:
            raise NoMoneyToWithdrawError(
                f"Недостаточно средств на счете '{account_from}'. "
                f"Доступно: {balance_from}, требуется: {value}"
            )

        # Выполнение перевода (транзакция)
        accounts[account_from] = balance_from - value
        accounts[account_to] = balance_to + value

    except (KeyError, ValueError, NoMoneyToWithdrawError) as e:
        # Перевыбрасываем известные исключения
        raise e
    except Exception as e:
        # Обработка непредвиденных ошибок
        raise PaymentError(f"Ошибка при переводе: {str(e)}")


if __name__ == "__main__":
    accounts = {
        "Василий Иванов": 100,
        "Екатерина Белых": 1500,
        "Михаил Лермонтов": 400
    }
    print_accounts(accounts)

    payment_info = {
        "account_from": "Екатерина Белых",
        "account_to": "Василий Иванов"
    }

    print("Перевод от {account_from} для {account_to}...".
          format(**payment_info))

    try:
        payment_info["value"] = int(input("Сумма = "))

        transfer_money(accounts, **payment_info)
        print("OK!")

    except ValueError as e:
        print(f"Ошибка ввода: {e}")
    except (KeyError, NoMoneyToWithdrawError, PaymentError) as e:
        print(f"Ошибка перевода: {e}")
    except Exception as e:
        print(f"Неожиданная ошибка: {e}")

    print_accounts(accounts)
```

Упражнение 10.
Напишите программу, которая будет суммировать все числа, введенные пользователем, игнорируя при этом нечисловой ввод. Выводите на экран текущую сумму чисел после каждого очередного ввода. Ввод пользователем значения, не являющегося числовым, должен приводить к появлению соответствующего предупреждения, после чего пользователю должно быть предложено ввести следующее число. Выход из программы будет осуществляться путем пропуска ввода. Удостоверьтесь, что ваша программа корректно обрабатывает целочисленные значения и числа с плавающей запятой.


```python
import re

def sum_numbers():
    total = 0
    print("Вводите числа для суммирования (Enter для выхода):")

    # шаблон: допускает целые и дробные числа, но без ведущих нулей (кроме самого нуля)
    number_pattern = re.compile(r'^[+-]?(0|[1-9]\d*)(\.\d+)?$')

    while True:
        user_input = input("Число: ").strip()

        # Выход при пустом вводе
        if user_input == "":
            break

        # Проверяем корректность ввода
        if not number_pattern.match(user_input):
            print("Ошибка: недопустимый формат числа (например: 5, -3.2, 0.5)")
            continue

        try:
            number = float(user_input)
            total += number
            print(f"Текущая сумма: {total}")
        except ValueError:
            print("Ошибка: не удалось преобразовать ввод в число.")

    print(f"\nИтоговая сумма: {total}")


# Запуск программы
sum_numbers()
```

```python
def convert_from_points(points):
    """Перевод суммы баллов в эквивалент и букву"""
    grade_scale = [
        (91, 100, 5, 'A', 'Отлично'),
        (84, 90, 4, 'B', 'Очень хорошо'),
        (74, 83, 4, 'C', 'Хорошо'),
        (68, 73, 3, 'D', 'Удовлетворительно'),
        (61, 67, 3, 'E', 'Посредственно'),
        (0, 60, 2, 'F', 'Неудовлетворительно')
    ]
    for low, high, num, letter, desc in grade_scale:
        if low <= points <= high:
            return f"{desc}: {letter} (эквивалент {num})"
    return "Значение вне диапазона (0–100)."


def convert_from_equivalent(eq):
    """Перевод числового эквивалента в диапазон баллов и букву"""
    grade_scale = {
        5: ('A', '91–100', 'Отлично'),
        4: ('B/C', '74–90', 'Очень хорошо / Хорошо'),
        3: ('D/E', '61–73', 'Удовлетворительно / Посредственно'),
        2: ('F', '0–60', 'Неудовлетворительно')
    }
    if eq in grade_scale:
        letter, rng, desc = grade_scale[eq]
        return f"{desc}: {letter} (диапазон баллов {rng})"
    return "Числовой эквивалент должен быть от 2 до 5."


def convert_from_letter(letter):
    """Перевод буквы в диапазон баллов и эквивалент"""
    letter = letter.upper()
    grade_scale = {
        'A': (5, '91–100', 'Отлично'),
        'B': (4, '84–90', 'Очень хорошо'),
        'C': (4, '74–83', 'Хорошо'),
        'D': (3, '68–73', 'Удовлетворительно'),
        'E': (3, '61–67', 'Посредственно'),
        'F': (2, '0–60', 'Неудовлетворительно')
    }
    if letter in grade_scale:
        num, rng, desc = grade_scale[letter]
        return f"{desc}: числовой эквивалент {num}, диапазон баллов {rng}"
    return "Недопустимая буква. Используйте A, B, C, D, E или F."


def main():
    print("=== Конвертер оценок ===")
    print("Выберите тип оценки для ввода:")
    print("1 — сумма баллов (0–100)")
    print("2 — числовой эквивалент (2–5)")
    print("3 — буквенная оценка (A–F)")
    print("Пустой ввод — выход\n")

    while True:
        choice = input("Тип оценки (1/2/3): ").strip()
        if not choice:
            print("Выход из программы.")
            break

        if choice == '1':
            try:
                value = int(input("Введите сумму баллов: ").strip())
                print(convert_from_points(value))
            except ValueError:
                print("Ошибка: нужно ввести число от 0 до 100.")

        elif choice == '2':
            try:
                value = int(input("Введите числовой эквивалент (2–5): ").strip())
                print(convert_from_equivalent(value))
            except ValueError:
                print("Ошибка: нужно ввести целое число от 2 до 5.")

        elif choice == '3':
            value = input("Введите буквенную оценку (A–F): ").strip()
            print(convert_from_letter(value))

        else:
            print("Неверный выбор. Введите 1, 2 или 3.")
        print()  # пустая строка для удобства


if __name__ == "__main__":
    main()
```
