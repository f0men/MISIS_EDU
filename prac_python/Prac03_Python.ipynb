Упражнение 1.
Дан список из чисел.
Определите их НОК (наименьшее общее кратное) и НОД (наибольший общий делитель).


```python
# Функции для двух чисел
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

def lcm(a, b):
    return a * b // gcd(a, b)

# Функции для списка чисел
def gcd_list(numbers):
    result = numbers[0]
    for num in numbers[1:]:
        result = gcd(result, num)
    return result

def lcm_list(numbers):
    result = numbers[0]
    for num in numbers[1:]:
        result = lcm(result, num)
    return result
```

```python
print(gcd(12, 18))
print(lcm(12, 18))

numbers = [12, 18, 24]
print(gcd_list(numbers))
print(lcm_list(numbers))
```

Упражнение 2.
Даны n предложений. Определите, сколько из них содержат хотя бы одну цифру.


```python
def count_sentences_with_digits(
    text)-> int:
    """Считает количество предложений с цифрами в тексте"""
    sentences = re.split(r'[.!?]', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    count = 0
    for sentence in sentences:
        if any(char.isdigit() for char in sentence):
            count += 1
    return count
```

```python
text = "Hello world! I have 2 apples. The year is 2023. No digits here. Price is 99 dollars"
result = count_sentences_with_digits(text)
print(f"Количество предложений с цифрами: {result}")
```

Упражнение 3.
Дана строка s и символ k. Реализуйте функцию, рисующую рамку из символа k вокруг данной строки, например:
**************
*Текст в рамке*
**************


```python
def write_frame(
    input_string : str,
    symb : str
    ) -> None:
    first_th_str = ((len(input_string)+2) * symb)
    second_str = symb + input_string + symb
    print(f'{first_th_str}\n{second_str}\n{first_th_str}')
```

```python
write_frame('Текст в рамке', '*')
```


Упражнение 4.
Для введенного предложения выведите статистику символ=количество. Регистр букв не учитывается.


```python
def counter_symb (
    text : str) -> dict:
    res = {}

    for sym in text.lower():
      if sym not in res and sym != ' ':
        res[f'{sym}'] = 1
      elif sym in res and sym != ' ':
        res[f'{sym}'] += 1
    return res
```

```python
counter_symb('Текст в рамке')
```

Упражнение 5.
Используя шифр Цезаря (достаточно только букв русского алфавита, знаки препинания не изменяются), зашифруйте, а затем расшифруйте введенную строку.


```python
def caesar_encrypt(text, shift):
    """Шифрует текст шифром Цезаря"""
    ru_upper = 'АБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯ'
    ru_lower = 'абвгдеёжзийклмнопрстуфхцчшщъыьэюя'

    result = []
    for char in text:
        if char in ru_upper:
            idx = (ru_upper.index(char) + shift) % len(ru_upper)
            result.append(ru_upper[idx])
        elif char in ru_lower:
            idx = (ru_lower.index(char) + shift) % len(ru_lower)
            result.append(ru_lower[idx])
        else:
            result.append(char)
    return ''.join(result)

def caesar_decrypt(text, shift):
    """Расшифровывает текст шифром Цезаря"""
    return caesar_encrypt(text, -shift)
```

```python
text = "Добрый вечер!"
shift = 5

print(f"Исходный: {text}")
encrypted = caesar_encrypt(text, shift)
print(f"Зашифрованный: {encrypted}")
decrypted = caesar_decrypt(encrypted, shift)
print(f"Расшифрованный: {decrypted}")
```

Упражнение 6.
Напишите функцию, которая принимает неограниченное количество числовых аргументов и возвращает кортеж из двух списков:
отрицательных значений (отсортирован по убыванию);
неотрицательных значений (отсортирован по возрастанию).


```python
def split_and_sort_numbers(*args):
    """
    Разделяет числа на отрицательные и неотрицательные и сортирует их

    Returns:
        tuple: (отрицательные_по_убыванию, неотрицательные_по_возрастанию)
    """
    negative = []
    non_negative = []

    for num in args:
        if num < 0:
            negative.append(num)
        else:
            non_negative.append(num)


    negative.sort(reverse=True)

    non_negative.sort()

    return (negative, non_negative)
```

```python
split_and_sort_numbers(3, 2, 1, 2, 10, 4, -10, -2, -3)
```

Упражнение 7.
Строка называется палиндромом, если она пишется одинаково в обоих направлениях. Например, палиндромами в английском языке являются слова «anna», «civic», «level», «hannah». Напишите программу, запрашивающую у пользователя строку и при помощи цикла определяющую, является ли она палиндромом.


```python
def is_palindrome(text):
    """Проверяет, является ли строка палиндромом"""
    # Приводим к нижнему регистру и убираем пробелы
    text = text.lower().replace(' ', '')
    length = len(text)

    # Проверяем симметричность
    for i in range(length // 2):
        if text[i] != text[length - i - 1]:
            return False
    return True
```

```python
print(is_palindrome('пока'))
print(is_palindrome('ала'))
```

Упражнение 8.
Вы загадываете число от 1 до 100 (включительно). Компьютер спрашивает у вас «Твое число равно, меньше или больше, чем число N?»,  где N — число, которое хочет проверить компьютер.
Вы отвечаете одним из трёх чисел:
1 — равно,
2 — больше,
3 — меньше.

Напишите программу, которая с помощью цепочки таких вопросов и ответов угадывает число.
Дополнительно: сделайте так, чтобы можно было гарантированно угадать число за семь попыток.


```python
def guess_number()-> None:
    low = 1
    high = 100

    while low <= high:
        guess = (low + high) // 2
        answer = int(input(f"Твое число равно, меньше или больше, чем число {guess}? (1-равно, 2-больше, 3-меньше): "))

        if answer == 1:
            print(f"Угадал! Число: {guess}")
            return
        elif answer == 2:
            low = guess + 1
        elif answer == 3:
            high = guess - 1

guess_number()
```


Упражнение 9.
Напишите программу, которая позволит пользователю преобразовывать числа из одной системы счисления в другую произвольным образом. Ваша программа должна поддерживать все системы счисления в диапазоне от 2 до 16 как для входных, так и для выходных данных. Если пользователь выберет систему с основанием, выходящим за границы допустимого, на экран должна быть выведена ошибка. Разделите код программы на несколько функций, включая функцию, конвертирующую число из произвольной системы счисления в десятичную, и обратную функцию, переводящую значение из десятичной системы в произвольную. В основной программе необходимо запросить у пользователя исходную систему счисления, целевую систему, а также число для преобразования.


```python
def to_decimal(number, base):
    """Конвертирует число из произвольной системы в десятичную"""
    digits = "0123456789ABCDEF"
    number = number.upper()
    decimal = 0

    for i, digit in enumerate(reversed(number)):
        if digits.index(digit) >= base:
            raise ValueError(f"Недопустимая цифра {digit} для системы с основанием {base}")
        decimal += digits.index(digit) * (base ** i)

    return decimal

def from_decimal(number, base):
    """Конвертирует число из десятичной системы в произвольную"""
    if number == 0:
        return "0"

    digits = "0123456789ABCDEF"
    result = ""

    while number > 0:
        result = digits[number % base] + result
        number //= base

    return result

def convert_number():
    """Основная функция конвертации"""
    try:
        source_base = int(input("Введите исходную систему счисления (2-16): "))
        target_base = int(input("Введите целевую систему счисления (2-16): "))
        number = input("Введите число для преобразования: ")

        if not (2 <= source_base <= 16) or not (2 <= target_base <= 16):
            print("Ошибка: системы счисления должны быть в диапазоне от 2 до 16")
            return

        decimal_num = to_decimal(number, source_base)

        result = from_decimal(decimal_num, target_base)

        print(f"Результат: {number} ({source_base}) = {result} ({target_base})")

    except ValueError as e:
        print(f"Ошибка: {e}")
    except Exception as e:
        print(f"Произошла ошибка: {e}")

convert_number()
```

Упражнение 10.

Магическими называются даты, в которых произведение дня и месяца составляет последние две цифры года. Например, 10 июня 1960 года – магическая дата, поскольку 10 ´ 6 = 60. Напишите функцию, определяющую, является ли введенная дата магической. Используйте написанную функцию в главной программе для отображения всех магических дат в XX ве­ке.


```python
def is_magic_date(
    day,
    month,
    year) -> bool:
    """Проверяет, является ли дата магической"""
    return day * month == year % 100

def find_magic_dates_20th_century():
    """Находит все магические даты в XX веке (1901-2000)"""
    days_in_month = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    for year in range(1901, 2001):
        for month in range(1, 13):
            # Проверяем високосный год для февраля
            if month == 2:
                if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
                    max_days = 29
                else:
                    max_days = 28
            else:
                max_days = days_in_month[month - 1]

            for day in range(1, max_days + 1):
                if is_magic_date(day, month, year):
                    print(f"{day:02d}.{month:02d}.{year}")

find_magic_dates_20th_century()
```
