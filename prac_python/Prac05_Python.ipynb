#№ 1 Задание

```python

class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return None

    def is_empty(self):
        return len(self.items) == 0


class TaskManager:
    def __init__(self):
        self.stack = Stack()

    def new_task(self, task, priority):
        """Добавляет таску по названию"""
        self.stack.push((task, priority))

    def remove_task(self, task):
        """Удаляет таску по названию"""
        temp_stack = Stack()
        found = False

        while not self.stack.is_empty():
            current_task, priority = self.stack.pop()
            if current_task == task and not found:
                found = True
            else:
                temp_stack.push((current_task, priority))

        while not temp_stack.is_empty():
            self.stack.push(temp_stack.pop())

        return found

    def __str__(self):
        tasks_by_priority = {}

        temp_stack = Stack()
        while not self.stack.is_empty():
            task, priority = self.stack.pop()
            temp_stack.push((task, priority))

            if priority not in tasks_by_priority:
                tasks_by_priority[priority] = []
            tasks_by_priority[priority].append(task)

        while not temp_stack.is_empty():
            self.stack.push(temp_stack.pop())

        result = []
        for priority in sorted(tasks_by_priority.keys()):
            tasks = tasks_by_priority[priority]
            result.append(f"{priority} {'; '.join(tasks)}")

        return '\n'.join(result)
```

```python
manager = TaskManager()
manager.new_task("сделать уборку", 4)
manager.new_task("помыть посуду", 4)
manager.new_task("отдохнуть", 1)
manager.new_task("поесть", 2)
manager.new_task("сдать дз", 2)
print("До удаления:")
print(manager)

print("\nПосле удаления 'поесть':")
manager.remove_task("поесть")
print(manager)
```

```python
class LRUCache:
    def __init__(self, capacity):
        self.capacity = capacity
        self.cache_dict = {}
        self.order = []

    @property
    def cache(self):
        if self.order:
            oldest_key = self.order[0]
            return (oldest_key, self.cache_dict[oldest_key])
        return None

    @cache.setter
    def cache(self, new_elem):
        key, value = new_elem

        if key in self.cache_dict:
            self.order.remove(key)
        elif len(self.cache_dict) >= self.capacity:
            oldest_key = self.order.pop(0)
            del self.cache_dict[oldest_key]

        self.cache_dict[key] = value
        self.order.append(key)

    def get(self, key):
        if key in self.cache_dict:
            self.order.remove(key)
            self.order.append(key)
            return self.cache_dict[key]
        return None

    def print_cache(self):
        print("LRU Cache:")
        for key in self.order:
            print(f"{key} : {self.cache_dict[key]}")
```

```python
cache = LRUCache(3)

cache.cache = ("key1", "value1")
cache.cache = ("key2", "value2")
cache.cache = ("key3", "value3")

cache.print_cache()

print(cache.get("key2"))

cache.cache = ("key4", "value4")

cache.print_cache()
```

#№ 3 Задание

```python
def cache_decorator(func):
    cache = {}

    def wrapper(n):
        if n not in cache:
            cache[n] = func(n)
        return cache[n]

    return wrapper

@cache_decorator
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
```

#№ 4 Задание

```python
class Cell:
    """Клетка игрового поля"""
    def __init__(self, number):
        self.number = number
        self.is_occupied = False
        self.value = ' '

    def __str__(self):
        return self.value if self.is_occupied else str(self.number)

class Board:
    """Игровое поле"""
    def __init__(self):
        self.cells = [Cell(i) for i in range(1, 10)]
        self.winning_combinations = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],  # горизонтальные
            [0, 3, 6], [1, 4, 7], [2, 5, 8],  # вертикальные
            [0, 4, 8], [2, 4, 6]              # диагональные
        ]

    def display(self):
        """Отображение игрового поля"""
        print("\n   |   |   ")
        print(f" {self.cells[0]} | {self.cells[1]} | {self.cells[2]} ")
        print("___|___|___")
        print("   |   |   ")
        print(f" {self.cells[3]} | {self.cells[4]} | {self.cells[5]} ")
        print("___|___|___")
        print("   |   |   ")
        print(f" {self.cells[6]} | {self.cells[7]} | {self.cells[8]} ")
        print("   |   |   \n")

    def make_move(self, cell_number, symbol):
        """Сделать ход на указанную клетку"""
        cell = self.cells[cell_number - 1]
        if not cell.is_occupied:
            cell.is_occupied = True
            cell.value = symbol
            return True
        return False

    def check_winner(self, symbol):
        """Проверка наличия победителя"""
        for combo in self.winning_combinations:
            if all(self.cells[i].value == symbol for i in combo):
                return True
        return False

    def is_full(self):
        """Проверка, заполнено ли все поле"""
        return all(cell.is_occupied for cell in self.cells)

class Player:
    """Игрок"""
    def __init__(self, name, symbol):
        self.name = name
        self.symbol = symbol

    def make_move(self, board):
        """Игрок делает ход"""
        while True:
            try:
                cell_number = int(input(f"{self.name}, выберите клетку (1-9): "))
                if 1 <= cell_number <= 9:
                    if board.make_move(cell_number, self.symbol):
                        return
                    else:
                        print("Эта клетка уже занята! Выберите другую.")
                else:
                    print("Введите число от 1 до 9!")
            except ValueError:
                print("Введите корректное число!")

class Game:
    """Управление игрой"""
    def __init__(self):
        self.board = Board()
        self.players = [
            Player("Игрок 1", "X"),
            Player("Игрок 2", "O")
        ]
        self.current_player_index = 0

    def switch_player(self):
        """Переключение на следующего игрока"""
        self.current_player_index = (self.current_player_index + 1) % 2

    def get_current_player(self):
        """Получение текущего игрока"""
        return self.players[self.current_player_index]

    def play(self):
        """Основной игровой цикл"""
        print("Добро пожаловать в Крестики-нолики!")
        print("Для выхода из игры введите 'q' в любое время")

        while True:
            self.board.display()
            current_player = self.get_current_player()

            # Проверка на выход
            move_input = input(f"{current_player.name}, выберите клетку (1-9) или 'q' для выхода: ")
            if move_input.lower() == 'q':
                print("Игра завершена.")
                break

            try:
                cell_number = int(move_input)
                if 1 <= cell_number <= 9:
                    if self.board.make_move(cell_number, current_player.symbol):
                        # Проверка победы
                        if self.board.check_winner(current_player.symbol):
                            self.board.display()
                            print(f"Поздравляем! {current_player.name} победил!")
                            break

                        # Проверка ничьей
                        if self.board.is_full():
                            self.board.display()
                            print("Ничья!")
                            break

                        self.switch_player()
                    else:
                        print("Эта клетка уже занята! Выберите другую.")
                else:
                    print("Введите число от 1 до 9!")
            except ValueError:
                print("Введите корректное число!")

        # Предложение сыграть еще раз
        play_again = input("Хотите сыграть еще раз? (y/n): ")
        if play_again.lower() == 'y':
            new_game = Game()
            new_game.play()

# Запуск игры
if __name__ == "__main__":
    game = Game()
    game.play()
```
