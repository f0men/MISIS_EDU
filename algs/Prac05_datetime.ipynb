```python
from datetime import datetime, timedelta
from dateutil.relativedelta import relativedelta
import time
```

1. Выведите текущую дату и время

```python
now = datetime.now()
print("Текущая дата и время:", now)
```

2. Преобразование строки в объект datetime

```python
date_string = "Feb 25 2020 4:20PM"
date_obj = datetime.strptime(date_string, "%b %d %Y %I:%M%p")
print("Преобразованная дата:", date_obj)
```

3. Вычтите неделю (7 дней) из заданной даты
Дано:

```python
iven_date = datetime(2020, 2, 25)
date_minus_week = iven_date - timedelta(days=7)
print("Дата минус 7 дней:", date_minus_week.date())
```

4. Выведите дату в следующем формате

```python
given_date = datetime(2020, 2, 25)
formatted_date = given_date.strftime("%A, %d %B %Y")
print("Форматированная дата:", formatted_date)
```

5. Найдите день недели для заданной даты

```python
given_date = datetime(2020, 7, 26)
weekday_name = given_date.strftime("%A")
print("День недели:", weekday_name)
```

6. Прибавьте к заданной дате неделю (7 дней) и 12 часов

```python
given_date = datetime(2020, 3, 22, 10, 0, 0)
new_date = given_date + timedelta(days=7, hours=12)
print("Дата + 7 дней и 12 часов:", new_date)
```

7. Выведите текущее время в миллисекундах

```python
milliseconds = int(time.time() * 1000)
print("Текущее время в миллисекундах:", milliseconds)
```

8. Преобразуйте следующее datetime в строку

```python
given_date = datetime(2020, 2, 25)
date_str = given_date.strftime("%Y-%m-%d %H:%M:%S")
print("Дата в строковом формате:", date_str)
```

9. Вычислите дату через 4 месяца от текущей даты

```python
given_date = datetime(2020, 2, 25).date()
future_date = given_date + relativedelta(months=4)
print("Дата через 4 месяца:", future_date)
```

10. Вычислите количество дней между двумя заданными датами

```python
date_1 = datetime(2020, 2, 25)
date_2 = datetime(2020, 9, 17)
delta_days = (date_2 - date_1).days
print("Количество дней между датами:", delta_days)
```
