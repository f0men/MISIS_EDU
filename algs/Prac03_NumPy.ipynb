13. Перемножить матрицы 5x3 и 3x2

```python
A = np.random.rand(5, 3)  # 5x3
B = np.random.rand(3, 2)  # 3x2

C = A @ B

print("Матрица A (5x3):\n", A)
print("\nМатрица B (3x2):\n", B)
print("\nРезультат C = A*B (5x2):\n", C)
```

14. Создать случайную матрицу 5x7 целых чисел от 1 до 100 и удалить [3,5] колонки

```python
matrix = np.random.randint(1, 101, (5, 7))

print("Исходная матрица:\n", matrix)

matrix_new = np.delete(matrix, [3, 5], axis=1)

print("\nМатрица после удаления колонок 3 и 5:\n", matrix_new)
```
