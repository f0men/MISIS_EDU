## Sort

1. Сортировка чисел по убыванию

```python
def bubble_sort(
    arr: list) -> list:

    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] < arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr

```

```python
numbers = [5, 2, 9, 1, 5]
bubble_sort(numbers)
```

2. Сортировка по длине

```python
def insertion_sort_by_length(
    arr: list) -> list:

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and len(arr[j]) > len(key):
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key
    return arr
```

```python
words = ["apple", "banana", "kiwi", "pear"]
insertion_sort_by_length(words)
```

3. Сортировка по сумме цифр

```python
def digit_sum(n):
    return sum(int(d) for d in str(abs(n)))
def sort_by_digit_sum(arr):
    n = len(arr)
    for i in range(n - 1):
        for j in range(n - i - 1):
            sum1, sum2 = digit_sum(arr[j]), digit_sum(arr[j + 1])
            if (sum1 > sum2) or (sum1 == sum2 and arr[j] > arr[j + 1]):
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return arr
```

```python
nums = [123, 45, 6, 789]
print(sort_by_digit_sum(nums))
```

4. Сортировка по частоте встречаемости

```python
def sort_by_frequency(
    words: list) -> list:
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1

    unique_words = []
    for w in words:
        if w not in unique_words:
            unique_words.append(w)

    n = len(unique_words)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if freq[unique_words[j]] < freq[unique_words[j + 1]]:
                unique_words[j], unique_words[j + 1] = unique_words[j + 1], unique_words[j]

    return unique_words

```

```python
words = ["apple", "banana", "apple", "cherry", "banana"]
sort_by_frequency(words)
```

5. Сортировка с сохранением исходных индексов

```python
def sort_with_indices(
    arr: list) -> list:
    indexed = [(i, arr[i]) for i in range(len(arr))]

    n = len(indexed)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if indexed[j][1] > indexed[j + 1][1]:
                indexed[j], indexed[j + 1] = indexed[j + 1], indexed[j]

    return indexed
```

```python
nums = [5, 2, 9, 1]
sort_with_indices(nums)
```

6. Сортировка с удалением дубликатов

```python
def unique_sorted(
    arr: list) -> list:

    unique = []
    for x in arr:
        if x not in unique:
            unique.append(x)

    n = len(unique)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if unique[j] > unique[j + 1]:
                unique[j], unique[j + 1] = unique[j + 1], unique[j]

    return unique
```

```python
nums = [5, 2, 9, 1, 5, 2]
print(unique_sorted(nums))
```

```python
x = np.linspace(0, 2 * np.pi, 200)


y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y_sin, color='red', linestyle='--', linewidth=2, label='sin(x)')
plt.plot(x, y_cos, color='blue', linestyle='-', linewidth=2, label='cos(x)')


plt.title("Графики функций sin(x) и cos(x)")
plt.xlabel("x")
plt.ylabel("Значение функции")
plt.grid(True)
plt.legend()

plt.show()
```

```python
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='k', alpha=0.9)

fig.colorbar(surf, shrink=0.5, aspect=10, label='z = sin(sqrt(x^2 + y^2))')

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title("3D поверхность: z = sin(sqrt(x^2 + y^2))")

plt.show()
```
